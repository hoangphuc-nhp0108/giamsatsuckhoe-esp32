```python
import cv2
import requests
import base64
import time

# ⚠️ BẮT BUỘC THAY THẾ URL NÀY
# Dán Public URL ngrok bạn nhận được từ Colab Server (Bước 1)
colab_url = "https://emelia-musicological-lael.ngrok-free.dev"

# --- CẤU HÌNH ---
# Điều chỉnh FPS để kiểm soát tốc độ gửi ảnh
FPS = 2
frame_delay = 1 / FPS

# Khởi tạo webcam (0 thường là webcam mặc định)
cap = cv2.VideoCapture(0)

if not cap.isOpened():
    print("❌ Lỗi: Không thể mở webcam. Kiểm tra driver hoặc đảm bảo không có ứng dụng nào khác đang sử dụng.")
    exit()

print(f"✅ Webcam đã mở. Bắt đầu gửi hình ảnh đến: {colab_url} với tốc độ {FPS} FPS...")
print("Nhấn phím 'q' để thoát.")

while True:
    ret, frame = cap.read()
    if not ret:
        break
    
    # Mã hóa hình ảnh thành chuỗi base64 (sử dụng định dạng JPEG để nén)
    _, img_encoded = cv2.imencode('.jpg', frame)
    img_base64 = base64.b64encode(img_encoded.tobytes()).decode('utf-8')

    # Gửi hình ảnh qua HTTP POST
    payload = {"image": img_base64}
    headers = {"Content-Type": "application/json"}
    
    try:
        response = requests.post(colab_url, json=payload, headers=headers, timeout=5)

        if response.status_code == 200:
            print(f"Gửi thành công! {time.strftime('%H:%M:%S')}")
        else:
            print(f"⚠️ Lỗi HTTP {response.status_code}: Không thể gửi hình ảnh. Phản hồi: {response.text}")

    except requests.exceptions.RequestException as e:
        print(f"❌ Lỗi kết nối (Timeout/Ngắt kết nối): {e}")

    # Hiển thị hình ảnh trên màn hình cục bộ (tùy chọn)
    cv2.imshow('Webcam Client (Nhan Q de thoat)', frame)
    
    # Thoát khi nhấn 'q'
    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

    # Đợi một khoảng thời gian trước khi gửi hình ảnh tiếp theo
    time.sleep(frame_delay)

# Giải phóng tài nguyên
cap.release()
cv2.destroyAllWindows()
print("Đã dừng chương trình và giải phóng webcam.")
```
